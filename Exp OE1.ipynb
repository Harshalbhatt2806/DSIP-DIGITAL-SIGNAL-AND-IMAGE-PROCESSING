```python
import numpy as np
import matplotlib.pyplot as plt

# Sample index
n = np.arange(-3, 7)

# Generate periodic sequence (1,2,3)
x = (n % 3) + 1

# Plot
plt.figure(figsize=(8,4))
markerline, stemlines, baseline = plt.stem(n, x)

plt.setp(markerline, markersize=8)
plt.setp(stemlines, linewidth=2)

plt.title('Discrete-Time Signal x[n]')
plt.xlabel('n')
plt.ylabel('x[n]')
plt.xticks(n)
plt.yticks([1, 2, 3])

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Sample index
n = np.arange(-3, 7)

# Create an array of zeros
x = np.zeros(len(n))

# Add values one by one
x[0] = 2
x[1] = 3
x[2] = 2
x[3] = 3
x[4] = 2
x[5] = 3
x[6] = 2
x[7] = 3
x[8] = 2
x[9] = 3

# Plot
plt.figure(figsize=(8,4))
plt.stem(n, x)

plt.title("Discrete-Time Signal x[n]")
plt.xlabel("n")
plt.ylabel("x[n]")
plt.xticks(n)
plt.yticks([1, 2, 3])

plt.show()
```
