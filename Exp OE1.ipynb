```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(-4, 9)

# Equation
# x = np.where(n % 3 == 1, 3, 2)
x = 3 - (n % 2)

plt.stem(n, x)
plt.xlabel("n")
plt.ylabel("x[n]")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Discrete-time index
n = np.arange(0, 6)

# Signal values
x = np.array([1, 2, 3, 3, 2, 1])

plt.stem(n, x, basefmt="k")
plt.xlabel("n")
plt.ylabel("x[n]")
plt.title("Discrete-Time Signal")
plt.grid(True)
plt.xlim(-1, 6)
plt.ylim(0, 4)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(-3, 4)

x = []

for i in n:
    if i < -2:
        x.append(0)
    elif -2 <= i < -1:
        x.append(-2)
    elif -1 <= i <= 1:
        x.append(2*i)
    elif 1 < i < 2:
        x.append(2)
    else:
        x.append(0)

plt.stem(n, x)
plt.xlabel('n')
plt.ylabel('x[n]')
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Range of n
n = np.arange(-2, 12)

# Unit step function
def u(n):
    return np.where(n >= 0, 1, 0)

# Signal
x = u(n) - u(n-3) - 5*u(n-7)

# Plot
plt.figure(figsize=(8,4))
plt.stem(n, x)

plt.xlabel('n')
plt.ylabel('x[n]')
plt.title(r'$x[n]=u(n)-u(n-3)-5u(n-7)$')
plt.grid(True)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Range of n
n = np.arange(-3, 4)

# Unit impulse function
def delta(n):
    return np.where(n == 0, 1, 0)

# Signal
x = delta(n) + 3*delta(n-1) + 5*delta(n+1)

# Plot
plt.figure(figsize=(8,4))
plt.stem(n, x)

plt.xlabel('n')
plt.ylabel('x[n]')
plt.title(r'$x[n]=\delta(n)+3\delta(n-1)+5\delta(n+1)$')
plt.grid(True)

plt.show()
```
