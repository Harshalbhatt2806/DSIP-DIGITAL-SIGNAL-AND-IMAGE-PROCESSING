```python

import cv2
import numpy as np
import matplotlib.pyplot as plt

# 1. Load the image in grayscale
image = cv2.imread( r"E:\MU\Sem-5\DSIP\LAB MANUAL\EXP\image\11.png",0)

if image is None:
    print("Error: Image not found! Check the file path.")
else:
    # 2. Convert grayscale image to binary
    _, binary = cv2.threshold(
        image, 127, 255, cv2.THRESH_BINARY
    )

    # 3. Define the structuring element
    # 1 = foreground, -1 = background, 0 = don't care
    pattern = np.array([
        [-1,  1, -1],
        [ 1,  1,  1],
        [-1,  1, -1]
    ], dtype=np.int8)

    # 4. Apply Hit-or-Miss Transformation
    result = cv2.morphologyEx(
        binary, cv2.MORPH_HITMISS, pattern
    )

    # 5. Display the original and result
    plt.figure(figsize=(10, 4))

    plt.subplot(1, 2, 1)
    plt.imshow(binary, cmap="gray")
    plt.title("Binary Image")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(result, cmap="gray")
    plt.title("Hit-or-Miss Result")
    plt.axis("off")

    plt.tight_layout()
    plt.show()
```
