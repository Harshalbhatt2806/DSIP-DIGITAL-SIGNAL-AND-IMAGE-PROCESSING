```python

import cv2
import numpy as np
import matplotlib.pyplot as plt

# 1. Load image in grayscale
image = cv2.imread( r"E:\MU\Sem-5\DSIP\LAB MANUAL\EXP\image\11.png",0)

if image is None:
    print("Error: Image not found! Check the file path.")
else:
    # 2. Define structuring element
    kernel = np.ones((5, 5), np.uint8)

    # 3. Perform Erosion
    erosion = cv2.erode(image, kernel, iterations=1)

    # 4. Perform Dilation
    dilation = cv2.dilate(image, kernel, iterations=1)

    # 5. Perform Opening
    opening = cv2.morphologyEx(
        image, cv2.MORPH_OPEN, kernel
    )

    # 6. Perform Closing
    closing = cv2.morphologyEx(
        image, cv2.MORPH_CLOSE, kernel
    )

    # 7. Display all results
    titles = [
        "Original Image",
        "Erosion",
        "Dilation",
        "Opening",
        "Closing"
    ]

    results = [
        image, erosion, dilation, opening, closing
    ]

    plt.figure(figsize=(14, 8))

    for i in range(5):
        plt.subplot(2, 3, i + 1)
        plt.imshow(results[i], cmap="gray")
        plt.title(titles[i])
        plt.axis("off")

    plt.tight_layout()
    plt.show()
```
